# Задание 1.

Загрузите из CSV-файла исходные данные о покупателях с веб-портала (до 15 строк). Проанализируйте возраст:

если больше 25 лет, то выводить «Старше 25»;
если меньше или равно, то выводить «Младше или равен 25».

Сделать через применение функции .apply pandas, изменить данные столбца DataFrame по условию.

```python
import pandas as pd


def load_clients(path, nrows=15):
    return pd.read_csv(path, nrows=nrows)


def age_category(age):
    if age > 25:
        return 'Старше 25'
    return 'Младше или равен 25'


def apply_age_categories(df):
    result = df.copy()
    result['age'] = result['age'].apply(age_category)
    return result


def filter_male_over_30(df):
    return df[(df['sex'] == 'male') & (df['age'] > 30)]


df = load_clients('web_clients_correct.csv')
df = apply_age_categories(df)
print(df['age'])
```
