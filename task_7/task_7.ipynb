# Задание 1.

Загрузите из CSV-файла исходные данные о покупателях с веб-портала (до 15 строк). При необходимости преобразуйте данные. С помощью Seaborn и Plotly сформируйте диаграмму разброса зависимости возраста клиента от расходов (scatterplot).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px


def load_clients(path, nrows=15):
    df = pd.read_csv(path)
    df = df.head(nrows).copy()
    df = df.dropna(subset=['age', 'bill'])
    return df


def plot_scatter_seaborn(df):
    plt.figure(figsize=(8, 5))
    sns.scatterplot(data=df, x='age', y='bill')
    plt.title('Зависимость расходов от возраста клиента')
    plt.xlabel('Возраст клиента')
    plt.ylabel('Расходы')
    plt.show()


def plot_scatter_plotly(df):
    fig = px.scatter(
        df,
        x='age',
        y='bill',
        title='Зависимость расходов от возраста клиента',
        labels={'age': 'Возраст клиента', 'bill': 'Расходы'}
    )
    fig.show()


def plot_regplot_seaborn(df):
    plt.figure(figsize=(8, 5))
    sns.regplot(data=df, x='age', y='bill')
    plt.title('Зависимость расходов от возраста клиента с линией регрессии')
    plt.xlabel('Возраст клиента')
    plt.ylabel('Расходы')
    plt.show()


df = load_clients('web_clients_correct.csv')
plot_scatter_seaborn(df)
```

Plotly:

```python
plot_scatter_plotly(df)
```
