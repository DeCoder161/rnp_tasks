# Задание 1.

Пользователь по команде «p» может узнать владельца документа по его номеру.

```python
documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}


def get_owner_by_number(doc_number, documents):
    for document in documents:
        if document['number'] == doc_number:
            return document['name']
    return None


def get_shelf_by_number(doc_number, directories):
    for shelf, numbers in directories.items():
        if doc_number in numbers:
            return shelf
    return None


def show_document_owner(documents):
    doc_number = input('Введите номер документа:\n')
    owner = get_owner_by_number(doc_number, documents)
    if owner:
        print(f'Владелец документа: {owner}')
    else:
        print('Владелец документа: владелец не найден')


def show_document_shelf(directories):
    doc_number = input('Введите номер документа:\n')
    shelf = get_shelf_by_number(doc_number, directories)
    if shelf:
        print(f'Документ хранится на полке: {shelf}')
    else:
        print('Документ не найден')


def main(documents, directories):
    while True:
        command = input('Введите команду:\n')
        if command == 'p':
            show_document_owner(documents)
        elif command == 's':
            show_document_shelf(directories)
        elif command == 'q':
            break
        else:
            print('Неизвестная команда')
```

Проверка для номера 10006:

```python
owner = get_owner_by_number('10006', documents)

if owner:
    print(f'Владелец документа: {owner}')
else:
    print('Владелец документа: владелец не найден')
```

Проверка для номера 1001:

```python
owner = get_owner_by_number('1001', documents)

if owner:
    print(f'Владелец документа: {owner}')
else:
    print('Владелец документа: владелец не найден')
```
