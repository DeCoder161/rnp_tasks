# Задание 1.

С помощью python смоделировать игру против лотерейного автомата типа «777».

Игрок платит 1 руб., после чего выпадает случайное целое число, равномерно распределенное от 0 до 999.
При некоторых значениях числа игрок получает выигрыш:

777: 200 руб.

999: 100 руб.

555: 50 руб.

333: 15 руб.

111: 10 руб.

*77: 5 руб.

**7: 3 руб.

*00: 2 руб.

**0: 1 руб.


(* – любая цифра)


Выгодна ли игра игроку?
Сколько в среднем приобретает или теряет игрок за одну игру?

```python
import numpy as np


def get_prize(number):
    if number == 777:
        return 200
    if number == 999:
        return 100
    if number == 555:
        return 50
    if number == 333:
        return 15
    if number == 111:
        return 10
    if number % 100 == 77:
        return 5
    if number % 10 == 7:
        return 3
    if number % 100 == 0:
        return 2
    if number % 10 == 0:
        return 1
    return 0


def simulate_games(n_games=1_000_000, seed=42):
    rng = np.random.default_rng(seed)
    numbers = rng.integers(0, 1000, size=n_games)
    prizes = np.array([get_prize(number) for number in numbers])

    total_spent = n_games
    total_won = prizes.sum()
    mean_result = (total_won - total_spent) / n_games

    return total_spent, total_won, mean_result


n_games = 1_000_000
total_spent, total_won, mean_result = simulate_games(n_games)

print(f'Игр: {n_games}')
print(f'Потрачено: {total_spent} руб.')
print(f'Выиграно: {total_won} руб.')
print(f'Средний результат за игру: {mean_result:.4f} руб.')
print(f'Выгодна ли игра игроку: {"да" if mean_result > 0 else "нет"}')
```

Игра игроку не выгодна: в среднем за одну игру он теряет около 0.2 руб.

# Задание 2.

Теоретически рассчитать средний выигрыш (проигрыш) и сравнить с результатами моделирования.

```python
def theoretical_mean():
    expected_prize = (
        1 * 200 +
        1 * 100 +
        1 * 50 +
        1 * 15 +
        1 * 10 +
        9 * 5 +
        90 * 3 +
        10 * 2 +
        90 * 1
    ) / 1000

    return expected_prize - 1


theory = theoretical_mean()

print(f'Теоретический средний результат: {theory:.4f} руб.')
print(f'Результат моделирования: {mean_result:.4f} руб.')
print(f'Разница: {abs(theory - mean_result):.4f} руб.')
```

Теоретически средний результат = −0.2 руб. за игру.

Моделирование на большом числе игр даёт значение близкое к −0.2, что совпадает с расчётом.
