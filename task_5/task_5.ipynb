# Задание 1.

Для каждого user_id в файле visit_log.csv определите третий столбец с категорией покупки. Используйте для работы файл purchase_log.txt.
Если покупка была, сам файл visit_log.csv изменять не надо.
Запишите в файл funnel.csv визиты из файла visit_log.csv, в которых были покупки с указанием категории.

содержимое visit_log.csv — не помещается в оперативную память компьютера, используйте только построчную обработку этого файла.

```python
import json

purchases = {}

with open('purchase_log.txt', encoding='utf-8') as f:
    for line in f:
        record = json.loads(line.strip())
        user_id = record['user_id']
        if user_id != 'user_id':
            purchases[user_id] = record['category']

with open('visit_log.csv', encoding='utf-8') as visits, \
     open('funnel.csv', 'w', encoding='utf-8') as funnel:
    funnel.write('user_id,source,category\n')
    next(visits)
    for line in visits:
        user_id, source = line.strip().split(',')
        if user_id in purchases:
            funnel.write(f'{user_id},{source},{purchases[user_id]}\n')

with open('funnel.csv', encoding='utf-8') as f:
    for i, line in enumerate(f):
        if i >= 3:
            break
        print(line.strip())
```
